## 1 Adding an Explanatory Variable to the Model
We've learned about the empty model: **outcome = mean + other stuff**
In today's lesson we'll build our first model in which we use an explanatory variable to explain variation in an outcome variable.

We use Sum of Squares to measure error from the empty model. The mean is the point that minimizes Sum of Squares in the outcome variable. But that still leaves a lot of error. Our goal in adding an explanatory variable is to reduce (or explain) some of that error.
