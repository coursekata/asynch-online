## 5 The Limitations of PRE

### 5.1 If the goal is to “improve” our model predictions, we could create a separate group for each of the 123 kids in this study. That way, each group could predict that student’s exact PS3 score. What do you think would happen to PRE? Why?

```r
# run this to see what happens

# this puts the kids into 120 tiny little groups
set.seed(1)
praisestudy <- praisestudy %>%
  mutate(group120 = ntile(PS3, 120),
         group120 = factor(group120, levels = sample(unique(group120))))
         
# this makes a big model based on that variable
big_model <- lm(PS3 ~ group120, data = praisestudy)

# let's add the predictions of big_model to this jitter plot
gf_jitter(PS3 ~ group120, data = praisestudy, height = 0, size=5) 
```
